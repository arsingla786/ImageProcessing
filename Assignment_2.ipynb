```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('lena.jpg')

height, width, channels = img.shape

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Calculate mean manually
for i in range(height):
    for j in range(width):
        B, G, R = img[i, j]
        gray[i, j] = (int(R) + int(G) + int(B)) / 3

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image - Mean Average")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('lena.jpg')

height, width, channels = img.shape

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Take weights from user
wr = float(input("Enter weight for R: "))
wg = float(input("Enter weight for G: "))
wb = float(input("Enter weight for B: "))

# Check weights
if wr < 0 or wr > 1 or wg < 0 or wg > 1 or wb < 0 or wb > 1:
    print("Weights must be between 0 and 1.")

elif abs(wr + wg + wb - 1) > 0.0001:
    print("Sum of weights must be equal to 1.")

else:
    for i in range(height):
        for j in range(width):
            B, G, R = img[i, j]

            gray[i, j] = (
                wr * int(R) +
                wg * int(G) +
                wb * int(B)
            )

    plt.imshow(gray, cmap='gray')
    plt.title("Grayscale Image - Weighted")
    plt.axis("off")
    plt.show()
```
