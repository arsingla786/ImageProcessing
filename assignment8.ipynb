```python
import numpy as np
import pandas as pd
import cv2
```

```python
img_path = '/content/Screenshot 2026-10-27 100414.png'
img = cv2.imread(img_path)
```

```python
import matplotlib.pyplot as plt

# Convert the image from BGR to grayscale
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Display the original and grayscale images side-by-side
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(gray_img, cmap='gray')
axes[1].set_title('Grayscale Image')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
gray_img.shape
```

```python
# Resize the grayscale image to 255x255
gray_img = cv2.resize(gray_img, (255, 255))
print("New shape:",gray_img.shape)

# Display the resized grayscale image
plt.figure(figsize=(6, 6))
plt.imshow(gray_img, cmap='gray')
```

```python
#noice function
def uniform_noice(image, mue, sigma):
    noise = np.random.normal(mue, sigma, image.shape)
    noisy_image = image + noise
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image
```

```python
uniform_noice_image = uniform_noice(gray_img, 0, 10)
plt.figure(figsize=(6, 6))
plt.imshow(uniform_noice_image, cmap='gray')
```

```python
def gaussian_noise(image, mean=0, sigma=15):
    noise = np.random.normal(mean, sigma, image.shape)
    noisy_image = image + noise
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image

# Apply Gaussian noise and display
gaussian_noise_image = gaussian_noise(gray_img, mean=0, sigma=15)
plt.figure(figsize=(6, 6))
plt.imshow(gaussian_noise_image, cmap='gray')
```

```python
def erlang_noise(image, a=2, b=10):
    # Erlang noise can be modeled using a Gamma distribution (shape parameter a, scale parameter b)
    noise = np.random.gamma(a, b, image.shape)
    noisy_image = image + noise
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image

# Apply Erlang noise and display
erlang_noisy_image = erlang_noise(gray_img, a=2, b=10)
plt.figure(figsize=(6, 6))
plt.imshow(erlang_noisy_image, cmap='gray')
```

```python
def exponential_noise(image, scale=10.0):
    # Generate exponential noise
    noise = np.random.exponential(scale, image.shape)
    noisy_image = image + noise
    # Keep pixels in [0, 255] boundary
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image

# Apply Exponential noise and display
exponential_noisy_image = exponential_noise(gray_img, scale=15.0)
plt.figure(figsize=(6, 6))
plt.imshow(exponential_noisy_image, cmap='gray')
```

```python
def rayleigh_noise(image, scale=15.0):
    # Generate Rayleigh noise
    noise = np.random.rayleigh(scale, image.shape)
    noisy_image = image + noise
    # Keep pixels within the valid [0, 255] range
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image

# Apply Rayleigh noise and display
rayleigh_noisy_image = rayleigh_noise(gray_img, scale=15.0)
plt.figure(figsize=(6, 6))
plt.imshow(rayleigh_noisy_image, cmap='gray')
```
